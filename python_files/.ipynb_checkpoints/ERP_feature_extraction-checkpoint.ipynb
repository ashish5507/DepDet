```python
import numpy as np

def extract_erp_features(epochs, condition):
    evoked = epochs[condition].average()
    times = evoked.times

    # P300: peak in 250-500ms window
    p300_window = (times >= 0.25) & (times <= 0.50)
    p300_amp = evoked.data[:, p300_window].mean(axis=0).max()
    p300_lat = times[p300_window][evoked.data[:, p300_window].mean(axis=0).argmax()]

    # P200: peak in 150-250ms window
    p200_window = (times >= 0.15) & (times <= 0.25)
    p200_amp = evoked.data[:, p200_window].mean(axis=0).max()

    return [p300_amp, p300_lat, p200_amp]
```
