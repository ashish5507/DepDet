```python
import librosa
import numpy as np

def extract_speech_features(audio_path):
    y, sr = librosa.load(audio_path, sr=16000)

    # MFCCs + deltas
    mfcc = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13)
    delta = librosa.feature.delta(mfcc)
    delta2 = librosa.feature.delta(mfcc, order=2)

    features = []
    for mat in [mfcc, delta, delta2]:
        features.extend(mat.mean(axis=1).tolist())
        features.extend(mat.std(axis=1).tolist())

    # Prosodic features
    f0, _, _ = librosa.pyin(y, fmin=50, fmax=300)
    rms = librosa.feature.rms(y=y)[0]
    zcr = librosa.feature.zero_crossing_rate(y)[0]

    features.extend([
        np.nanmean(f0), np.nanstd(f0),
        rms.mean(), rms.std(),
        zcr.mean()
    ])

    return np.array(features)  # shape: (83,)
```
